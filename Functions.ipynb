#Theory Questions:

1. What is the difference between a function and a method in Python?

*   Function: A block of reusable code defined using def or lambda, independent of any object.
*   Method: A function that belongs to an object (typically defined inside a class).

Example:

```python
'''
def greet(): return "Hello"     # Function
"hello".upper()                 # Method
'''
```

2. Explain the concept of function arguments and parameters in Python.

*   Parameters: Variables listed in a function definition.
*   Arguments: Actual values passed to the function.

Example:

```python
'''
def add(a, b): return a + b  # a, b are parameters
add(3, 4)                    # 3, 4 are arguments
'''
```

5. What are iterators in Python and how do they differ from iterables?


*   Iterable: Object you can loop over (like list, tuple, string).
*   Iterator: Object that remembers the state during iteration, accessed using next().

Example:

```python
'''
lst = [1, 2, 3]      # Iterable
it = iter(lst)       # Iterator
next(it)
'''
```

6. Explain the concept of generators in Python and how they are defined.

   -> Generators are special functions that yield values one at a time using yield keyword.

   Example:

```python
'''
def count_up(n):
    for i in range(n):
        yield i
'''
```

7. What are the advantages of using generators over regular functions?


*   Memory-efficient (don't store entire list)
*   Support lazy evaluation (yield items one-by-one)
*   Simpler syntax for creating iterators



8. What is a lambda function in Python and when is it typically used?

  -> A lambda function is an anonymous function defined with the lambda keyword, used for short, simple functions.

  Example:

```python
'''
add = lambda a, b: a + b
add(2, 3)  # Output: 5
'''
```

9. Explain the purpose and usage of the map() function in Python.

  -> map() applies a given function to each item of an iterable and returns a map object (can be converted to list).

  Example:

```python
'''
nums = [1, 2, 3]
doubled = list(map(lambda x: x*2, nums))  # Output: [2, 4, 6]
'''
```

10. What is the difference between map(), reduce(), and filter() functions in Python?


*   map(): Transforms elements
*   filter(): Filters elements based on condition
*   reduce(): Combines elements into a single value (requires functools)

Example:

```python
'''
from functools import reduce
nums = [1, 2, 3, 4]

list(map(lambda x: x*2, nums))       # [2, 4, 6, 8]
list(filter(lambda x: x % 2 == 0, nums))  # [2, 4]
reduce(lambda x, y: x + y, nums)     # 10
'''
```

```python
# 5.Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

'''
class Fibonacci:
    def __init__(self, n):
        self.n, self.a, self.b, self.count = n, 0, 1, 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return self.a
'''
```

```python
# 6.Write a generator function in Python that yields the powers of 2 up to a given exponent.

'''
def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i
'''
```

```python
# 7.Implement a generator function that reads a file line by line and yields each line as a string.

'''
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()
'''
```

```python
# 8.Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

'''
tuples = [(1, 3), (2, 1), (4, 2)]
sorted_tuples = sorted(tuples, key=lambda x: x[1])
'''
```

```python
# 9.Write a Python program that uses map() to convert a list of temperatures from Celsius to Fahrenheit.

'''
celsius = [0, 20, 30, 40]
fahrenheit = list(map(lambda c: (9/5)*c + 32, celsius))
'''
```

```python
# 10.Create a Python program that uses filter() to remove all the vowels from a given string.

'''
def remove_vowels(s):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', s))
'''
```
